```python
from sklearn import preprocessing
import numpy as np
```

```python
# Carga de datos
X_train = np.array([
    [1., -1., 2.],
    [2., 0., 0.],
    [0., 1., -1.]
])

X_train
```

```python
# Limpieza de datos: estandarización

standardizer = preprocessing.StandardScaler()

# Estandarizamos los datos de entrenamiento
X_train_std = standardizer.fit_transform(X_train)

print(X_train_std)
```

```python

X_train_2 = np.array([
    [1., -1., 2.],
    [2., 0., 0.],
    [0., 1., -1.]
])

X_train_2
```

```python
# Aplicamos los métodos "fit" y "transform" por separado

f = preprocessing.StandardScaler()

# Ajustamos el estandarizador con los datos de entrenamiento
f = f.fit(X_train)

# Transformamos los datos
X_train_2_std = f.transform(X_train_2)

print(X_train_2_std)
```

```python
# Comprobar si las matrices estandarizadas son iguales

print(np.array_equal(X_train_std, X_train_2_std))
```
